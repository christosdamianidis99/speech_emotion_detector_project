```python
from pathlib import Path
import librosa
import numpy as np
import matplotlib.pyplot as plt

PROJECT_ROOT = Path(r"C:\Code\emotion-detector-project")

RAVDESS_ROOT = Path(r"C:\Code\Python\data\RAVDESS")
CREMAD_ROOT  = Path(r"C:\Code\Python\data\CREMA-D\AudioWAV")
IEMOCAP_ROOT = Path(r"C:\Code\Python\data\IEMOCAP\IEMOCAP_full_release")

print("RAVDESS exists:", RAVDESS_ROOT.exists())
print("CREMA-D exists:", CREMAD_ROOT.exists())
print("IEMOCAP exists:", IEMOCAP_ROOT.exists())
```

```python
# Utility to show a random example
import random

def show_example(path_list, title):
    path = random.choice(path_list)
    print(f"{title} example file:\n{path}\n")

    y, sr = librosa.load(path, sr=None)  # keep original sample rate
    duration = len(y) / sr
    print(f"Sample rate: {sr} Hz, duration: {duration:.2f} seconds")

    plt.figure(figsize=(10, 3))
    librosa.display.waveshow(y, sr=sr)
    plt.title(f"{title} waveform")
    plt.xlabel("Time (s)")
    plt.ylabel("Amplitude")
    plt.tight_layout()
    plt.show()
```

```python
ravdess_files = sorted(RAVDESS_ROOT.glob("**/*.wav"))
print("RAVDESS file count:", len(ravdess_files))

show_example(ravdess_files, "RAVDESS")
```

```python
cremad_files = sorted(CREMAD_ROOT.glob("*.wav"))
print("CREMA-D file count:", len(cremad_files))

show_example(cremad_files, "CREMA-D")
```

```python
iemocap_files = sorted(IEMOCAP_ROOT.glob("**/*.wav"))
print("IEMOCAP file count:", len(iemocap_files))

show_example(iemocap_files, "IEMOCAP")
```
