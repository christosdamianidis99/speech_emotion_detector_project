```python
import pandas as pd
from sklearn.model_selection import train_test_split

all_df = pd.read_csv("metadata_step2.csv")  
```

```python
# 1) First split: train (70%), temp (30%)
train_df, temp_df = train_test_split(
    all_df,
    test_size=0.30,
    stratify=all_df["emotion"],
    random_state=42,
)

# 2) Second split: validation (15%), test (15%)
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["emotion"],
    random_state=42,
)

# Assign split labels
train_df["split"] = "train"
val_df["split"] = "val"
test_df["split"] = "test"

# Combine into final metadata
final_df = pd.concat([train_df, val_df, test_df], ignore_index=True)

final_df.to_csv("metadata_final.csv", index=False)

final_df.head()
```

```python
print(final_df["split"].value_counts())

print("\nEmotion distribution by split:\n")
print(pd.crosstab(final_df["emotion"], final_df["split"]))
```
